```python
n = np.linspace(1, 6, 6)
I = (1/12)*(1**3)*0.95
A = 1 * 0.95
def long_innspent(L, E, rho):
    return ((n*np.pi - (np.pi/2))/L)*np.sqrt(E/rho)
def trans_innspent(L, E, rho):
    return (((np.pi*n)/L)**2) * np.sqrt((E*I)/(rho*A))

longditudanal_søyle = long_innspent(5, 50, 2)
transverse_søyle = trans_innspent(5, 50, 2)
longditudanal_bjelke = long_innspent(10, 50, 2)
transverse_bjelke = trans_innspent(10, 50, 2)

print(longditudanal_søyle)
print(transverse_søyle)
print(longditudanal_bjelke)
print(transverse_bjelke)
```
